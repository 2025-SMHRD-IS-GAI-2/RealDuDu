```python
import oracledb
import datetime 

# ==============================================
# ★ [핵심 수정] 구버전 DB 접속을 위한 'Thick Mode' 활성화
# ==============================================
try:
    # PC에 설치된 오라클 클라이언트를 찾아서 연결합니다.
    oracledb.init_oracle_client()
    print("✅ 오라클 클라이언트 모드(Thick Mode) 초기화 성공")
except Exception as e:
    print("⚠️ 초기화 경고 (일단 무시하고 아래 접속 시도해보세요):", e)
    # 만약 여기서 "Library not found" 같은 에러가 나서 접속이 안 되면, 
    # 오라클이 설치된 폴더 경로를 직접 적어줘야 합니다. (맨 아래 설명 참고)

# ==============================================
# 2. DB 연결 정보 (본인 걸로 확인!)
# ==============================================
db_ip = "project-db-campus.smhrd.com" 
db_port = 1524 
db_sid = "xe" 
db_id = "campus_25IS_GA2_p2_4" 
db_pw = "smhrd4" 

# ==============================================
# 3. 연결 함수
# ==============================================
def get_db_connection():
    dsn = oracledb.makedsn(db_ip, db_port, sid=db_sid)
    try:
        conn = oracledb.connect(user=db_id, password=db_pw, dsn=dsn)
        print(f"✅ DB 접속 성공!")
        return conn
    except Exception as e:
        print(f"❌ 접속 실패: {e}")
        return None

# ==============================================
# 4. 실행 및 INSERT 테스트
# ==============================================
conn = get_db_connection()

if conn:
    cursor = conn.cursor()
    
    # [주의] 사진에 테이블 이름은 안 보여서 'YOUR_TABLE_NAME'이라고 뒀습니다.
    # 실제 테이블 이름(예: HELMET_LOG)으로 꼭 바꿔주세요!
    sql = """
        INSERT INTO TB_RIDE (
            RIDE_ID, USER_ID, KICKBOARD_ID, START_DT, END_DT, 
            HELMET_TM, NOHELMET_CNT, SCORE_CG, RIDE_ST
        ) 
        VALUES (:1, :2, :3, :4, :5, :6, :7, :8, :9)
    """
    
    current_time = datetime.datetime.now()

    # 테이블 컬럼 타입(NUMBER, VARCHAR2)에 맞춰 데이터 준비
    data = (
        "JUPITER_RIDE000",   # RIDE_ID
        "USER010",     # USER_ID
        "DD010",    # KICKBOARD_ID
        current_time, # START_DT
        current_time, # END_DT
        12.5,         # HELMET_TM
        3,            # NOHELMET_CNT
        10,           # SCORE_CG
        "E"           # RIDE_ST
    )

    try:
        cursor.execute(sql, data)
        conn.commit()  # ★저장 확정
        print("🎉 데이터 입력 성공! DB 확인해보세요.")
    except Exception as e:
        print("💥 쿼리 실행 에러:", e)
        conn.rollback()
    
    # conn.close()
```

```python
# 루프 오류를 해결 해야함.
from fastapi import FastAPI
import uvicorn
import nest_asyncio

# 주피터의 기존 루프에 얹혀가기 위한 패치
nest_asyncio.apply()

app = FastAPI()

# 1. 기본 접속 테스트
@app.get("/")
def read_root():
    return {"message": "Hello! 여기는 파이썬 서버입니다."}

# 2. 자바 요청 처리
@app.get("/helmet-check")
def helmet_check(kickboard_id: str):
    print(f"📡 자바에서 요청 도착! 킥보드 ID: {kickboard_id}")
    return {"result": "성공", "status": "헬멧 검사 완료"}

# ==========================================
# 👇 여기가 중요합니다! (코드가 바뀌었습니다)
# ==========================================
if __name__ == "__main__":
    # uvicorn.run() 대신 Server 객체를 만들어서 비동기(await)로 실행해야 합니다.
    config = uvicorn.Config(app=app, host="127.0.0.1", port=8000)
    server = uvicorn.Server(config)
    
    # 주피터 셀은 이미 비동기 환경이라 'await'를 바로 쓸 수 있습니다.
    await server.serve()
```

```python
# 로컬주소로 해서 민하 컴퓨터에서 테스트 불가능!! ip 주소를 보내줘서 제대로 실험
import uvicorn
import nest_asyncio
from fastapi import FastAPI

nest_asyncio.apply()
app = FastAPI()

@app.get("/")
def read_root():
    return {"message": "Hello! 팀원 접속 성공!"}

@app.get("/helmet-check")
def helmet_check(kickboard_id: str):
    print(f"📡 팀원(자바)에게서 요청 도착! ID: {kickboard_id}")
    return {"result": "성공", "status": "헬멧 검사 완료"}

if __name__ == "__main__":
    # ★ host를 "0.0.0.0"으로 바꿔야 다른 컴퓨터가 들어올 수 있어요!
    # 포트는 8001번 유지
    config = uvicorn.Config(app=app, host="0.0.0.0", port=8001)
    server = uvicorn.Server(config)
    
    print("🚀 서버가 열렸습니다! 팀원에게 IP를 알려주세요.")
    await server.serve()
```

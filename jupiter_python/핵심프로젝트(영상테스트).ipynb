# 주피터노트북 영상 파일 만들기

```python
video_path = './data/minha.mp4'
cap = cv2.VideoCapture(video_path)
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
fps = cap.get(cv2.CAP_PROP_FPS)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
out = cv2.VideoWriter('result_last.mp4', fourcc, fps, (width, height))

print("녹화 및 분석 시작... (Q를 누르면 중단됩니다)")

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break
    results = model(frame)
    annotated_frame = results[0].plot()
    cv2.imshow("YOLOv11 Inference", annotated_frame)
    out.write(annotated_frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
cap.release()
out.release()
cv2.destroyAllWindows()
print("저장 완료 : result_last.mp4")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. 파일 불러오기 및 공백 제거
results_path = './data/results.csv' 
df = pd.read_csv(results_path)
df.columns = df.columns.str.strip()

# ★ 중요: 여기서 출력되는 리스트를 꼭 확인하세요!
print("현재 내 파일의 컬럼명들:")
print(df.columns.tolist()) 
print("-" * 30)

# 2. 내 파일에 맞는 이름으로 매칭하기
# 아래 ['이름'] 부분들을 위에서 출력된 리스트 중 가장 비슷한 이름으로 바꿔주세요.

plt.figure(figsize=(12, 5))

# [왼쪽 그래프: Loss]
plt.subplot(1, 2, 1)
# 보통 'train/box_loss' 또는 'train/box_om' 등으로 되어있을 겁니다.
loss_col = [c for c in df.columns if 'box_loss' in c and 'train' in c][0]
val_loss_col = [c for c in df.columns if 'box_loss' in c and 'val' in c][0]
plt.plot(df[loss_col], label='Train Loss')
plt.plot(df[val_loss_col], label='Val Loss')
plt.title('Loss Curve')
plt.legend()

# [오른쪽 그래프: mAP]
plt.subplot(1, 2, 2)
# 'metrics/mAP_0.5'가 없다면 'mAP_0.5' 또는 'metrics/mAP_0.5(B)'일 확률이 높습니다.
# 가장 비슷한 컬럼을 자동으로 찾는 코드입니다.
map_col = [c for c in df.columns if 'mAP_0.5' in c and '0.5:0.95' not in c][0]
plt.plot(df[map_col], label='mAP 0.5')
plt.title('mAP Performance')
plt.legend()

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. 파일 불러오기 및 공백 제거
results_path = './data/results.csv' 
df = pd.read_csv(results_path)
df.columns = df.columns.str.strip()

# 2. 정확한 컬럼명 찾기 (에러 방지용)
# Loss 관련 컬럼 찾기
train_loss = [c for c in df.columns if 'train' in c and 'box' in c][0]
val_loss = [c for c in df.columns if 'val' in c and 'box' in c][0]

# mAP 관련 컬럼 찾기 (mAP_0.5 또는 mAP50 또는 mAP50(B) 등 모두 검색)
map_candidates = [c for c in df.columns if ('mAP' in c and '50' in c) or ('mAP' in c and '0.5' in c)]
# 0.5:0.95 처럼 범위로 된 것은 제외하고 가장 기본 mAP만 선택
map_col_list = [c for c in map_candidates if '95' not in c]

if not map_col_list:
    print("❌ mAP 컬럼을 찾지 못했습니다. 아래 리스트 중 정확도와 관련된 이름을 알려주세요!")
    print(df.columns.tolist())
else:
    map_col = map_col_list[0]
    print(f"✅ 사용된 mAP 컬럼: {map_col}")

    # 3. 그래프 그리기
    plt.figure(figsize=(12, 5))

    # [왼쪽: Loss]
    plt.subplot(1, 2, 1)
    plt.plot(df[train_loss], label='Train Loss')
    plt.plot(df[val_loss], label='Val Loss')
    plt.title('Loss Curve')
    plt.legend()

    # [오른쪽: mAP]
    plt.subplot(1, 2, 2)
    plt.plot(df[map_col], label='mAP 50', color='red')
    plt.title('mAP Performance')
    plt.legend()

    plt.show()
```
